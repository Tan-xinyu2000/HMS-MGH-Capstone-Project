#### Mutual Information 

```python

# Calculate mutual information scores
mutual_info = mutual_info_regression(X, y)

# Convert scores to a pandas Series and associate with column names
mutual_info = pd.Series(mutual_info)
mutual_info.index = X.columns

# Sort by mutual information in descending order
sorted_mi = mutual_info.sort_values(ascending=False)

# Plot the scores
sorted_mi.plot.bar(figsize=(20, 8))

# Select top 50 features
top_50_features = sorted_mi.head(10).index

print(top_50_features)
```

#### K best 

```python
# https://scikit-learn.org/stable/modules/feature_selection.html
# For regression, can use r_regression, f_regression, mutual_info_regression
```

```python
# r_regression 
kbest_selector_r = SelectKBest(r_regression, k=20) # k keeps amount of most significant variables 
X_new = kbest_selector_r.fit_transform(X, y)

selected_features_r = np.where(kbest_selector_r.get_support())[0]
print("Selected feature indices:", selected_features_r)
selected_column_names_r = X.columns[kbest_selector_r.get_support()].tolist()
print("Selected column names:", selected_column_names_r)
```

```python
# f_regression

kbest_selector_f = SelectKBest(f_regression, k=20) # k keeps amount of most significant variables 
X_new = kbest_selector_f.fit_transform(X, y)

selected_features_f = np.where(kbest_selector_f.get_support())[0]
print("Selected feature indices:", selected_features_f)
selected_column_names_f = X.columns[kbest_selector_f.get_support()].tolist()
print("Selected column names:", selected_column_names_f)
```

```python
# mutual_info_regression 

kbest_selector_mutual = SelectKBest(mutual_info_regression, k=20) # k keeps amount of most significant variables 
X_new = kbest_selector_mutual.fit_transform(X, y)

selected_features_mutual = np.where(kbest_selector_mutual.get_support())[0]
print("Selected feature indices:", selected_features_mutual)
selected_column_names_mutual = X.columns[kbest_selector_mutual.get_support()].tolist()
print("Selected column names:", selected_column_names_mutual)
```
